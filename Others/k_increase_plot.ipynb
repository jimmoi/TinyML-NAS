```python
import matplotlib.pyplot as plt
import math

def plot_nc(k_val, max_c):
    c_values = [0]
    n_values = [k_val]
    
    for c in range(1, max_c + 1):
        # The term (2 - sum_{i=1}^{c-1} 2^-i) simplifies to (1 + 2^{-(c-1)})
        multiplier = 1 + pow(2, -(c - 1))
        next_n = math.ceil(multiplier * n_values[-1])
        
        c_values.append(c)
        n_values.append(next_n)
    
    # Plotting the results
    plt.figure(figsize=(10, 6))
    plt.plot(c_values, n_values, marker='o', linestyle='-', color='b')
    plt.title(f'Plot of $n_c$ vs $c$ (Starting value $k={k_val}$)')
    plt.xlabel('c')
    plt.ylabel('$n_c$')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()

# Execute the function for c from 0 to 30 with k=1
plot_nc(k_val=4, max_c=5)
```
