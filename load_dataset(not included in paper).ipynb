```python
from pathlib import Path
import shutil

print("Cloning GIT...")
# clone repo (still needs shell, Git is external)
!git clone https://github.com/teavanist/MNIST-JPG.git
print()

print("Unzipping with shutil...")

zip_path = Path("MNIST-JPG/MNIST Dataset JPG format.zip")
extract_dir = Path(".")

shutil.unpack_archive(zip_path, extract_dir)

# Move to Datasets
src = Path("MNIST Dataset JPG format")
dst = Path("../Datasets/MNIST").resolve()

if dst.exists():
    raise FileExistsError("Destination already exists!")

shutil.move(src, dst)

print("Moved to:", dst)
```

```python
import shutil
import os
from pathlib import Path
import stat

def force_remove(func, path, exc_info):
    # Remove read-only attribute and retry
    os.chmod(path, stat.S_IWRITE)
    func(path)

path = Path("MNIST-JPG")

if path.exists():
    shutil.rmtree(path, onerror=force_remove)
    print(f"Removed: {path}")
else:
    print(f"Not found: {path}")
```
