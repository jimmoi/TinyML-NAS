### Flowers-4

```python
import kagglehub
import shutil
from pathlib import Path

# Download latest version
src_path = kagglehub.dataset_download("l3llff/flowers")
src_path = str(src_path)
src_path = Path(src_path +'\\flowers\\')

dst_path = Path("../Datasets/Flowers-4")

# Delete do not use classes
remove_classes = [
    "astilbe",
    "bellflower",
    "black_eyed_susan",
    "calendula",
    "california_poppy",
    "carnation",
    "common_daisy",
    "coreopsis",
    "daffodil",
    "rose",
    "sunflower",
    "water_lily",
]
for cls in remove_classes:
    path = src_path / cls
    if path.exists():
        shutil.rmtree(path)
        print(f"Removed: {cls}")
    else:
        print(f"Not found: {cls}")

# Move (destination must NOT exist)
if dst_path.exists():
    raise FileExistsError("Destination folder already exists!")

shutil.move(str(src_path), str(dst_path))

print("Dataset moved to:", dst_path.resolve())
```

### Animals-3


```python
import kagglehub
import shutil
from pathlib import Path

# Download latest version
src_path = kagglehub.dataset_download("alessiocorrado99/animals10")
print(src_path)
src_path = str(src_path)
src_path = Path(src_path +'\\raw-img\\')

dst_path = Path("../Datasets/Animals-3")

# Delete do not use classes
remove_classes = ["cane", "elefante", "gatto", "mucca", "pecora", "scoiattolo", "ragno"]
for cls in remove_classes:
    path = src_path / cls
    if path.exists():
        shutil.rmtree(path)
        print(f"Removed: {cls}")
    else:
        print(f"Not found: {cls}")
        
# base = Path("Animals-3")  # change if needed

# Move (destination must NOT exist)
if dst_path.exists():
    raise FileExistsError("Destination folder already exists!")

shutil.move(str(src_path), str(dst_path))

print("Dataset moved to:", dst_path.resolve())
print()

# Rename folders
translate = {"cavallo": "horse", 
            "farfalla": "butterfly", 
            "gallina": "hen"}
print("Renaming folders")
for old_name, new_name in translate.items():
    src = dst_path / old_name
    dst = dst_path / new_name
    if not src.exists():
        print(f"Skip (not found): {old_name}")
        continue
    if dst.exists():
        print(f"Skip (target exists): {new_name}")
        continue
    src.rename(dst)
    print(f"Renamed: {old_name} → {new_name}")
print()
```
